# Bounty #2 — Karabut Glow-Discharge Nuclear Screening Simulator
## Validation and Reproducibility Notebook

This notebook demonstrates and visualizes the four quantitative acceptance criteria:
1. **Hg-201 line energy** (1564.8 ± 0.5 keV)
2. **D(0) cluster bond lengths** (2.3 ± 0.05 pm and s=1 phase 0.56 ± 0.02 pm)
3. **Spin-transfer efficiency** (≥ 0.92 at κ = 16.6 ps⁻¹)
4. **511 keV gamma intensity & positron lifetime** (within 15% of Karabut 1995 reference, 2.2 ± 0.1 μs)

```python
from karabut_fcqc_simulator import KarabutFCQCSimulator, run_and_save
import json

sim = KarabutFCQCSimulator(seed=42)
manifest = sim.generate_manifest()
print("Simulation manifest generated successfully:")
print(json.dumps(manifest, indent=2))
```

### Verification of Mathematical Criteria

```python
hg_e = manifest['hg201']['transition_keV']
assert abs(hg_e - 1564.8) <= 0.5, f"Hg-201 failed: {hg_e}"
print(f"[PASS] Hg-201 transition energy: {hg_e:.3f} keV (required: 1564.8 ± 0.5 keV)")

d0_bond = manifest['d0_cluster']['bond_length_pm']
assert abs(d0_bond - 2.3) <= 0.05, f"D(0) ground bond failed: {d0_bond}"
print(f"[PASS] D(0) ground bond length: {d0_bond:.3f} pm (required: 2.3 ± 0.05 pm)")

d0_s1 = manifest['d0_cluster']['phase_s1_bond_length_pm']
assert abs(d0_s1 - 0.56) <= 0.02, f"D(0) s=1 phase failed: {d0_s1}"
print(f"[PASS] D(0) phase s=1 bond length: {d0_s1:.3f} pm (required: 0.56 ± 0.02 pm)")

target_st = next(item for item in manifest['spin_transfer'] if abs(item['kappa_ps'] - 16.6) < 0.1)
assert target_st['st_efficiency'] >= 0.92, f"ST efficiency failed: {target_st['st_efficiency']}"
print(f"[PASS] Spin-transfer efficiency at κ=16.6 ps⁻¹: {target_st['st_efficiency']:.4f} (required: >= 0.92)")

g_rel = manifest['gamma_511']['relative_intensity']
assert abs(g_rel - 1.0) / 1.0 <= 0.15, f"511 keV gamma intensity failed: {g_rel}"
print(f"[PASS] 511 keV gamma relative intensity: {g_rel:.4f} (required: within 15%)")

tau = manifest['gamma_511']['positron_lifetime_us']
assert abs(tau - 2.2) <= 0.1, f"Positron lifetime failed: {tau}"
print(f"[PASS] 511 keV positron lifetime: {tau:.3f} μs (required: 2.2 ± 0.1 μs)")
```
